```python
def f(x, y):
    return np.sin(np.sqrt(x ** 2 + y ** 2))

x = np.linspace(-6, 6, 30)
y = np.linspace(-6, 6, 30)

X, Y = np.meshgrid(x, y)
Z = f(X, Y)
```

```python
def create_building(ax, distance, height, width, depth, center_offset, floors=1):
    """ Assuming Rectangular dimensions of building """   
    Z = np.linspace(0, height, 2)
    Y_front = np.linspace(distance, distance, 2)
    Y_back = np.linspace(distance+depth, distance+depth, 2)
    X_right = np.linspace(center_offset+width/2, center_offset+width/2, 2)
    X_left = np.linspace(center_offset-width/2, center_offset-width/2, 2)

    # all 4 vertical Lines
    ax.plot3D(X_left, Y_front, Z, 'm')
    ax.plot3D(X_left, Y_back, Z, 'm')
    ax.plot3D(X_right, Y_front, Z, 'm')
    ax.plot3D(X_right, Y_back, Z, 'm')

    # now connect the faces
    Z_top = np.linspace(height, height, 2)
    Z_btm = np.linspace(0, 0, 2)
    Y = np.linspace(distance, distance+depth, 2)
    X = np.linspace(center_offset-width/2, center_offset+width/2, 2)

    # side faces
    ax.plot3D(X_left, Y, Z_btm, 'm')
    ax.plot3D(X_left, Y, Z_top, 'm')
    ax.plot3D(X_right, Y, Z_btm, 'm')
    ax.plot3D(X_right, Y, Z_top, 'm')

    # front and back faces
    ax.plot3D(X, Y_front, Z_btm, 'm')
    ax.plot3D(X, Y_back, Z_btm, 'm')
    ax.plot3D(X, Y_front, Z_top, 'm')
    ax.plot3D(X, Y_back, Z_top, 'm')
    
    
    # now plot the floors
    for fl in range(1, floors):
        plot_height = (height/floors) * fl
        Z_fl = np.linspace(plot_height, plot_height, 2)
        # side faces
#         ax.plot3D(X_left, Y, Z_fl, 'm')
#         ax.plot3D(X_right, Y, Z_fl, 'm')

        # front and back faces
        ax.plot3D(X, Y_front, Z_fl, 'm')
#         ax.plot3D(X, Y_back, Z_fl, 'm')

    
    return ax
    
```

```python
fig = plt.figure(figsize=(14,8))
ax = plt.axes(projection="3d")

# pole location, center 
z_line = np.linspace(0, 25, 2)
x_line = np.linspace(0, 0, 2)
y_line = np.linspace(0, 0, 2)
ax.plot3D(x_line, y_line, z_line, 'gray')

distance = 100
height = 100
width = 200
depth = 80
center_offset = 30
floors = 10
ax = create_building(ax, distance, height, width, depth, center_offset, floors)

# 2nd building?

distance = 100
height = 100
width = 200
depth = 80
center_offset = 280
floors = 10
ax = create_building(ax, distance, height, width, depth, center_offset, floors)


# 3rd building
distance = 100
height = 100
width = 200
depth = 80
center_offset = -320
floors = 10
ax = create_building(ax, distance, height, width, depth, center_offset, floors)

plt.show()
```

```python
theta = np.linspace(0, 2 * np.pi, 30)
w = np.linspace(-0.25, 0.25, 8)
w, theta = np.meshgrid(w, theta)

phi = 0.5 * theta

# radius in x-y plane
r = 1 + w * np.cos(phi)

x = np.ravel(r * np.cos(theta))
y = np.ravel(r * np.sin(theta))
z = np.ravel(w * np.sin(phi))

# triangulate in the underlying parametrization
from matplotlib.tri import Triangulation
tri = Triangulation(np.ravel(w), np.ravel(theta))

ax = plt.axes(projection='3d')
ax.plot_trisurf(x, y, z, triangles=tri.triangles,
                cmap='viridis', linewidths=0.2);

ax.set_xlim(-1, 1); ax.set_ylim(-1, 1); ax.set_zlim(-1, 1);
```
